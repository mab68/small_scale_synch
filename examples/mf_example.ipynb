```python
NSIDE = 512
LMAX = 3*NSIDE - 1

ells = np.arange(LMAX+1)
alpha = 8./3.
L = 10.
target_cl = np.sqrt(ells**2 + L**2)**(-alpha)
target_cl[np.isnan(target_cl)|np.isinf(target_cl)] = 1e-15

LOW_C, HIGH_C = 0.001, 5.0
LOW_LAMB, HIGH_LAMB = 0.001, 0.5

t_delta_params = {
    'cl_target': target_cl,
    'cl_noise': target_cl,
    'c':LOW_C,
    'lambda':LOW_LAMB,
    'anisotropy':'isotropic',
}
e_delta_params = {
    'cl_target': target_cl,
    'cl_noise':target_cl,
    'c':LOW_C,
    'lambda':LOW_LAMB,
    'anisotropy':'isotropic',
}
b_delta_params = {
    'cl_target':target_cl,
    'cl_noise':target_cl,
    'c':LOW_C,
    'lambda':LOW_LAMB,
    'anisotropy':'isotropic',
}

cl_tt, cl_ee, cl_bb, cl_te = 1.*target_cl, 0.75*target_cl, 0.2*target_cl, 0.3*target_cl

i_delta1, _, _ = synthesize.make_iqu(
    LMAX, NSIDE, t_delta_params, e_delta_params, b_delta_params, cl_tt, cl_ee, cl_bb, cl_te)
```
